```python
from ib_insync import *
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
ib = IB()
```

```python
util.startLoop()
ib.disconnect()
ib.connect('127.0.0.1', 7497, clientId = 100)
```

```python
ib.run()
```

```python
def extract_info(accounts, tag):
    for acc in accounts:
        if tag == acc.tag:
            return acc.value
    return None
```

```python
account = ib.accountSummary()
print("Net worth:", extract_info(account, "NetLiquidation"))
print("Cash:", extract_info(account, "TotalCashValue"))
```

```python
bars = ib.reqHistoricalData(
            # contract=Crypto('BTC', "PAXOS", "USD"), #, primaryExchange='NASDAQ'),
            contract=Stock('GOOG', 'SMART', 'USD', primaryExchange='NASDAQ'),
            endDateTime='',
            durationStr=f'{60 * 60 * 1} S',
            barSizeSetting='1 min',
            whatToShow='MIDPOINT',
            useRTH=True, 
            formatDate = 2
        )
```

```python
util.df(bars)
```

```python
from trading import TradingBot

# Define bot parameters
args = {
    # Stochastic RSI Attributes
    'rsi_period': 14,
    'stochastic_period': 9,

    # Bollinger Band Attributes
    'simple_moving_average_period': 21,
    'bollinger_band_standard_deviation': 1.5,

    # Buy and Sell Attributes
    'ticker': 'COST',  
    'stochastic_upper_band': 70,
    'stochastic_lower_band': 30,
    'time_look_back':  f'1 D', 
    'asset_interval': '30 secs', 
    'buy_quantity': 1, 
    'show_times': 500,
    'mode': 'trade', 
    'limit_percent': 0.015,
    'initial_cash': 1e4,
}

# Create and run bot
bot = TradingBot(ib, args)
while True:
    bot.backtest()
    ib.sleep(30)
    # break
```
